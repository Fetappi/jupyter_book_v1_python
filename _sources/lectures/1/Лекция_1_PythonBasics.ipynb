# Основы Python

## Лексемы
Python разбивает каждую логическую строку на ряд элементарных лексических
компонентов, так называемых лексем. Каждая лексема соответствует подстроке
логической строки. К числу обычных типов лексем относятся идентификаторы, ключевые
слова, операторы, разделители и литералы.

Идентификатор - это имя, используемое для задания переменной, функции,
класса, модуля или другого объекта. Идентификатор должен начинаться с буквы
символа подчеркивания (_).

В Python имеются ключевые слова - идентификаторы, зарезервированные для специального использования.

````python
import keyword
import cmd

print(len(keyword.kwlist))
cli = cmd.Cmd()
cli.columnize(keyword.kwlist, displaywidth=50)  # type: ignore
````

Литерал - это значение (число, строка или контейнер), указанное непосредственно в программе.

## Типы данных

В python всё — от булевых значений и чисел до функций, классов и файлов — представляется в виде объекта.

Каждый объект имеет тип. Тип определяет атрибуты и элементы (если таковые имеются), а также допускается ли изменение объекта.

У каждого такого объекта есть три главные характеристики: идентификатор (id), тип (type) и значение.

````python
print(id(5))
print(type(5))
print((5).bit_length())
````

### Числовые типы

#### `int` - целые числа

Верхняя граница для целочисленных литералов не определена, и они могут
принимать сколь угодно большие значения.

````python
import sys
print(sys.getsizeof(1))
print(sys.getsizeof(2**50))
````

#### `float` - числа с плавающей точкой

Числа с плавающей точкой реализованы по тому же стандарту, что и в C/C++ (IEEE 754), представлены 64-х битным типом **float** (соответствует **double** в C/C++)

````python
e = 2.71828
print(sys.getsizeof(e))  # 8 bytes + 16 bytes of CPython object overhead
````

#### `complex` - комплексные числа

Комплексное число состоит из двух чисел с плавающей точкой, представляющих соответственно его действительную и мнимую части.

Для записи комплексного числа в качестве мнимой единицы можно использовать строчную или прописную букву **j**.

````python
a = 1 + 2j
b = 1 + 2J
c = complex(1, 2)
print(a == b == c)
print(a.real, a.imag)
````

#### `bool` - булевый тип

Встроенный тип `bool` - это подкласс типа `int`. Он может принимать только
два значения - `True` или `False`, которые имеют строковые представления 'True'
и 'False', но в действительности являются значениями 1 и 0 соответственно. Булевы
значения возвращаются операторами сравнения, а также рядом встроенных функций.

В Python любое значение может служить показателем логической истинности или
ложности: `True` или `False`. Любое ненулевое значение или непустой контейнер (на
пример, строка, кортеж, список, множество или словарь) является истинным значе
нием (эквивалентными значению `True`). 0 (любого числового типа), `None` и пустые
контейнеры являются ложными значениями (эквивалентными значению `False`).

Функция `bool(х)` может принимать любой аргумент **х**. Она возвращает значение
`True`, если **х** представляет истинное значение, и `False`, если **х** представляет ложное
значение. Однако ороший стиль программирования на Python требует избегать таких вызовов, если в этом нет особой необходимости.

#### Объект `None` типа `NoneType`

В python есть специальный тип данных `NoneType`, который может принимать единственно возможное значение `None`.

Объект `None` обозначает нулевой объект. Он не имеет методов и других атрибутов. Объект `None` можно использовать в качестве заместителя в тех случаях, когда требуется ссылка, но вам безразлично, на какой объект она указывает, или когда требуется указать на отсутствие какого-либо объекта. Если функция не содержит инструкций `return`, возвращающих другие значения, то она возвращает объект `None`.

````python
print(type(None))
````

### Контейнеры

#### Контейнер
Контейнер (`Container`) — это объект, который позволяет проверить, содержится ли в нем элемент (поддерживает оператор `in`).
Строго говоря, тип считается контейнером, если у него перегружен метод `__contains__`, что и позволяет его использовать справа от ключевого слова `in`.

Должно выполняться:
```python
x in A
```

#### Итерируемый объект
Итерируемый объект (`Iterable`) — объект, по которому можно пробежаться в цикле (iterate over).

Должно выполняться:
```python
for x in A
```

#### Объект ограниченной длины
Объект ограниченной длины (`Sized`) - объект, у которго можно спросить количество элементов (длина, размер, size) функцией `len`.

Должно выполняться:
```python
len(A)
```

#### Коллекция
Коллекция (`Collection`) — это объект, который одновременно является контейнером, итерируемым объектом и объектом ограниченный длины.

````python
from collections.abc import Collection, Container, Iterable, Sized

print(issubclass(Collection, Container))
print(issubclass(Collection, Iterable))
print(issubclass(Collection, Sized))
````

По способу получения доступа к элементам можно выделить два наиболее широких класса коллекций — последовательности (Sequence) и отображения (Mapping), хотя ещё существуют множества (Set) и другие.

#### Последовательность
Последовательность (`Sequence`) - это упорядоченная коллекция элементов, индексируемых
с помощью целых чисел, т.е. у любой последовательности можно спросить её i-й элемент. Индексация последовательностей в python всегда начинается с 0 (zero based).

* `str` - строки,
* `list` - списки,
* `tuple` - кортежи,
* `bytearray` и `bytes` - изменяемый и неизменяемый массивы байтов.

Должно выполняться:
```python
A[i]
```

````python
from collections.abc import Collection, Sequence, Mapping

print(issubclass(Sequence, Collection))
print(issubclass(Mapping, Collection))
````

#### Общие для последовательностей операции

| Операция     | Описание                   | Свойство |
| ------------ | -------------------------- | --- |
| `len(s)`     | Длина последовательности **s**.| Sized |
| `x in s`     | True, если в s есть элемент равный x. False иначе.| Container |
| `for x in s` | Перебор элементов в **s**.| Iterable |
| `s + t`      | Конкатенация **s** и **t**. Новый объект. | Iterable |
| `s * n` или `n * s` |Эквивалентно добавлению **s** к самому себе **n** раз. Новый объект. | Iterable |
| `s.index(x)` | Индекс первого совпадения **x** с элементом в **s**. | Iterable |
| `s.count(x)` | Сколько раз **x** встречается в **s**. | Iterable |
| `max(s)`     | Наибольший элемент в **s**. | Iterable |
| `sorted(s)`  | Список элементов **s** в отсортированном порядке. | Iterable |

Последние две операции возможны, если элементы **s** можно сравнить друг с другом на предмет большинства и равенства.

#### `list` - списки

**Список** - это изменяемая упорядоченная последовательность произвольных объектов. Это наиболее близкий аналог к массиву из C/C++. Существенным отличием является возможность хранения в одном и том же списке объекты разных типов.

Списки - изменяемые объекты: элементам и срезам списков можно присваивать
значения или удалять их. Элементами списка могут быть произвольные объекты, которые не обязательно должны быть одного типа. Срезы списков являются списками.

````python
l = [1, 2, 3, 4]              # Список из элементов одного типа
l = [1, "hello", int, True]   # Список из элементов разного типа
l = [0]                       # Список из одного элемента
l = []                        # Пустой список
l = list()                    # Тоже пустой список
l = list("ego")               # Литеральный список ["e", "g", "o"]
l = [x**2 for x in range(5)]  # Создание списка с помощью спискового включения
l = [0] * 5                   # Создание списка с помощью оператора умножения
l = [[0] * 3] * 2             # Но вот так делать не стоит
````

Методы типа `list`

Методы, которые не изменяют объект (немутирующие методы).

| Операция     | Описание |
| ------------ | -------- |
| `l.count(x)`  | Считает, сколько раз элемент **x** встречается в **l**.|
| `l.index(x[, start[, end]])` |  Возвращает индекс первого появления элемента **x**. |
| `l.copy()` | Создаёт копию списка. |

Методы, которые изменяют объект (мутирующие методы).

| Операция     | Описание |
| ------------ | -------- |
| `l[i] = x`   | Изменяет **i**-й элемент на **x**. |
| `l.append(x)` | Добавляет элемент в конец списка. |
| `l.extend(t)` или `l += t` | Расширяет список, добавляя все элементы итерируемого объекта в конце списка. |
| `l.insert(i, x)` | Вставляет элемент **x** на указанную позицию (индекс) **i**. `l.insert(0, x)` вставляет **x** в самое начало списка. |
| `l *= n` | То же, что и `l = l * n`, но не создаёт новый список, а расширяет существующий. |
| `l.clear()` | Полностью очищает список, удаляя все элементы. |
| `l.pop()` или `l.pop(i)` | Удаляет элемент по индексу **i** и возвращает его. Если индекс не указан, удаляет последний элемент. |
| `l.remove(x)` | Удаляет первое встретившееся значение **x**. |
| `l.sort()` или  `l.sort(key=None, reverse=False)` | Cортирует элементы списка на месте. |
| `l.reverse()` |  Обращает список на месте. |
| `l[i:j] = t` | Срез с **i** по **j** заменяется элементами из **t**. |
| `del l[i:j]` | То же самое, что и `l[i:j] = []`. |
| `l[i:j:k] = t` | Элементы среза с **i** по **j** с шагом **k** заменяются на элементы из **t**, но `len(t)` должно быть равно `len(l[i])`. |
| `del l[i:j:k]` | Удаляет элементы этого среза из списка. |

Пустой список приводится к `False`. Любой не пустой список приводится к `True`. Иногда это используется для проверки на пустоту списка. Но лучше всегда делать эту проверку явной: `len(l) != 0`.

````python
l = []
print(bool(l))
print(len(l) != 0)
````

#### `tuple` - кортежи

**Кортеж** - это неизменяемая упорядоченная последовательность произвольных объектов.

Кортежи - неизменяемые объекты, поэтому при попытке присвоить значение элементу или срезу кортежа или удалить его генерируется исключение. Однако элементы кортежа могут быть изменяемыми, что может приводить к неожиданным результатам.

````python
t = (1, 2, 3, 4)             # Кортеж из элементов одного типа
t = (1, "hello", int, True)  # Кортеж из элементов разного типа
t = (0,)                     # Кортеж из одного элемента: требуется замыкающая запятая
t = ()                       # Пустой кортеж
t = tuple()                  # Тоже пустой кортеж
````

Методы типа `tuple`

Так как кортеж является неизменяемым (immutable) типом данных, то для него существуют только немутирующие методы.

| Операция | Описание |
| -------- | -------- |
| `t.count(x)` | Считает, сколько раз элемент **x** встречается в **t**. |
| `t.index(value, [start, [stop]])` | Находит индекс первого появления элемента **x** в **t**. Можно задать диапазон поиска через необязательные параметры start (начало) и stop (конец). |

Очень часто кортежи используются для того, чтобы упаковать несколько объектов, передать их как единое целое и затем распаковать эти объекты обратно.

````python
d, m = divmod(11, 3)  # Распаковка кортежа, возвращаюмего функцией
````

#### "Сырые" строки и f-строки

**Сырые строки** (raw strings) - это текстовые литералы, в которых символ обратного слэша (`\`) воспринимается как обычный символ, а не как управляющий или экранирующий. Сырая строка создается добавлением префикса `r` или `R` перед открывающей кавычкой.

````python
s = "\\hello"
print(s)
s = r"\\hello"
print(s)
````

Основные способы форматирования строк — это современные f-строки, метод `.format()` и старый оператор `%`.

**f-строки** - это удобный способ подставления переменных в строки и их форматирования с помощью префикса `f` (или `F`) и фигурных скобок `{}`.

Метод `.format()` - Это более старый универсальный метод, который поддерживается во всех версиях Python 3. Значения передаются в скобки через метод `.format()`, по порядку или по именам.

````python
x = 1
y = 3.1415
z = 10000
s = "{} {:.2f} {:.0E}".format(x, y, z)
print(s)
s = f"{x} {y:.2f} {z:.0E}"
print(s)
````

````python
s = {1, 2, 3, 4}             # Множество из элементов одного типа
s = {1, "hello", int, True}  # Множество из элементов разного типа
s = {0}                      # Множество из одного элемента
s = set()                    # Пустое множество
s = {}                       # А это не пустое множество, это пустой словарь!
````

#### `dict` - словари

**Отображение** (Mapping) - это произвольная коллекция объектов, индексируемых с помощью почти произвольных значений - ключей. Как и множества, отображения
являются изменяемыми и, в отличие от последовательностей, не обязательно упорядоченными.

Python предоставляет один встроенный тип отображений: словарь.

Ключи словаря могут иметь различные типы, но они должны быть *хешируемыми*. Значениями словаря могут быть произвольные объекты любого типа. Элементами словаря являются пары "ключ-значение".

````python
d = {"x": 23, "y": 45}            # Словарь с ключами типа str
d = dict(x=23, y=45)              # Другим способом

d = {1: "ab", 2: "xy"}            # Словарь с ключами типа int
d = dict([(1, "ab"), (2, "xy")])  # Другим способом

d = {"x": 23, 2: "ab"}            # Словарь с ключами разных типов
d = dict([("x", 23), (2, "ab")])  # Другим способом

d = {}                            # Пустой словарь
d = dict()                        # Пустой словарь

# Создание словаря с помощью метода fromkeys
d = dict.fromkeys('ego', 42)  # = {"e": 42, "g": 42, "o": 42}
d = dict.fromkeys([1, 2])     # = {1: None, 2: None}
````

Методы типа `dict`

Методы, которые не изменяют объект (немутирующие методы).

| Операция | Описание |
| -------- | -------- |
| `d.copy()` | |
| `d.get(key[, default])`| Возвращает значение `d[key]`, если **key** является ключом в **d**; в противном случае возвращает **default** (по умолчанию `None`). |
| `d.keys()` | Возвращает представление всех ключей словаря. |
| `d.items()` | Возвращает представление всех пар (ключ, значение) в виде списка кортежей.|
| `d.values()`| Возвращает представление всех значений словаря. |

Методы, которые изменяют объект (мутирующие методы).

| Операция | Описание |
| -------- | -------- |
| `d.clear()` | Удаляет все элементы из **d**. |
| `d.update(t)`| Обновляет словарь парами из **t** (другого словаря или итерабельного объекта с парой ключ-значение), перезаписывая существующие ключи. |
| `d.pop(key[, default])` | Удаляет ключ key и возвращает его значение. Если ключа нет и default не задан, вызывает ошибку. |
| `d.popitem()` | Удаляет и возвращает последнюю пару (ключ, значение) (в Python 3.7+ последнюю добавленную).|
| `d.setdefault(key[, default])` | Возвращает значение по ключу **key**. Если ключа нет, добавляет его в словарь со значением **default**. |

## Управляющие инструкции

Поток управления - это очередность выполнения кода программы. Поток управления
программы на языке Python зависит от условных инструкций, инструкций циклов и вызовов функций.

#### Инструкция `if`

Составная инструкция `if`, включающая предложения `if`, `elif` и `else`, обеспечивает условное выполнение блоков инструкций.

````python
x = -2
if x < 0:
    print ( 'x - отрицательное число')
elif x % 2:
     print('x - нечетное положительное число')
else:
    print('x - четное неотрицательное число')
````

#### Инструкция `while`

Инструкция `while` обеспечивает многократное выполнение инструкции или блока инструкций, управляемых условным выражением.

Инструкция while может включать предложение `else`, а также инструкции `break` и `continue`.

````python
i = 0
while i < 5:
    i += 1
print(i)
````

#### Инструкция `for`

Инструкция `for` обеспечивает многократное выполнение инструкции или блока инструкций, управляемых итерируемым выражением.

Инструкция `for` может включать предложение `else`, а также инструкции `break` и `continue`.

Далее инструкция `for` будет обсуждаться в разделе Итераторы.

#### Инструкция `break`

Инструкцию `break` разрешается использовать только в теле цикла.
Выполнение инструкции `break` приводит к преждевременному выходу из цикла.
Если цикл вложен в другие циклы, то данная инструкция прерывает только ближайший вложенный цикл.

````python
while True:
    n = int(input("Введите число:"))
    if n == 0:
        break
````

#### Инструкция `continue`

Инструкция `continue` может встречаться лишь в теле цикла. Выполнение ин
струкции `continue` прерывает выполнение текущей итерации тела цикла и передает
управление в начало цикла.

````python
for s in ["ok", "ok", None, "ok"]:
    if s is None:
        continue
    print(s, end=" ")  # Этот код теперь не нужно делать вложенным в условную обработку
````

#### Предложение `else` в инструкциях циклов

Инструкции `while` и `for` могут включать необязательное завершающее предложение `else`.
Инструкция или блок, следующие за ключевым словом `else`, выполняются, если цикл завершился естественным образом (вследствие исчерпания итератора цикла `for` или когда условие цикла `while` приобрело ложное значение), но не тогда, когда цикл завершается преждевременно (вследствие выполнения инструкции `break` либо `return` или возникновения исключения).

````python
l = [100, 101, 102, 103]
for i in l:
    # 'g' == chr(103)
    if chr(i) == "g": break
else:
    print("g not found!")
````

#### Инструкция `pass`

Тело составной инструкции Python не может быть пустым; оно должно содержать
хотя бы одну инструкцию. В подобных случаях, когда синтаксис требует наличия
инструкции, можно использовать инструкцию-заполнитель `pass`, которая не выполняет никаких действий.

````python
x = 0
if x == 0:
    pass
````

#### Инструкция `try` и `raise`

Python поддерживает обработку исключений с помощью инструкции `try`, включающей предложения `try`, `except`, `finally` и `else`. Ваш код может явно генерировать исключение с помощью инструкции raise. Когда генерируется исключение, обычный поток управления программы прерывается, и Python выполняет поиск подходящего обработчика исключений.

````python
try:
    n = int(input("Не вводите 0!"))
    if n == 0:
        print("Exception raised!")
        raise RuntimeError
except RuntimeError:
    print("Exception caught!")
    #raise RuntimeError
else:
    print("No exception!")
finally:
    # Will run, no matter what happens
    print("Bye.")
````

#### Инструкция `with`

Иногда можно улучшить читаемость кода за счет использования инструкции
`with` вместо инструкции `try`/`finally`.

С этой инструкцией связано понятие контекстного менеджера.
Контекстный менеджер в Python — это объект, который автоматически управляет ресурсами (например, открывает и закрывает файлы или соединения с базами данных) с помощью конструкции `with`.

````python
with open('file.txt', 'r') as f:
    text = f.read()
# Здесь файл закроется автоматически
````

## Итераторы

Итератор - это любой объект `i`, обеспечивающий возможность выполнения вызова `next(i)` (т.е. должен определить метод `__next__`). Метод `next(i)` возвращает следующий элемент итератора `i` или, в случае исчерпания элементов итератора, генерирует исключение `Stopiteration`.

Получить стандартный итератор по коллекции можно встроенной функцией `iter`.

````python
for letter in "ego":
    print(letter, end=", ")
````

Совет: не изменяйте изменяемые объекты при выполнении цикла по их злементам.

````python
l = [1, 2, 3]
iterator = iter(l)
print(next(iterator))
print(next(iterator))
print(next(iterator))
print(next(iterator))
````

Цикл `for` использует итераторы для прохода по итерируемым объектам и сам вызывает функцию `next` до тех пор, пока не встретит исключение `StopIteration`.

Инструкция
```python
for х in с:
    ...
```
в точности эквивалентна следующему коду:
```python
temporary_iterator = iter(c)
while True:
    try: х = next(temporary_iterator)
    except Stopiteration: break
    ...
```

### Встроенные итераторы

#### `range`

Фукция `range` позволяет сделать обход последовательности целых чисел.

Функция `range(start, stop, step)` возвращает список целых чисел, которые заключены в диапазоне от `start` (включительно) до `stop` (исключая это значение) и принимают значения, отсчитываемые от начального значения с заданным шагом. Если величина шага меньше 0, то отсчет значений в направлении их
убывания ведется от `start` до `stop`.

````python
for i in range(0, 6, 2):
    print(i, end=" ")
````

#### `enumerate`

Функция `enumerate`возвращает “нумерующий итератор”, т.е. итератор, который помимо следующего элемента контейнера возвращает ещё и его номер.

````python
word = "ego"
for i, letter in enumerate(word):
    print(i, letter)
````

#### `reversed`

Функция `reversed` создаёт итератор, который пробегает по коллекции в обратном порядке.

````python
word = "ego"
for letter in reversed(word):
    print(letter, end=" ")
````

#### `zip`

Функция `zip` создаёт итератор по нескольким итерируемым объектам, т.е. позволяет пробежаться по нескольким контейнерам одновременно.

````python
a = "AB"
b = [1, 2]
c = ("a", "b")
for x, y, z in zip(a, b, c):
    print(x, y, z)
````

## Функции

Функция - это группа инструкций, выполняемых по запросу. Запрос выполнения функции называют вызовом функции.
Функции, определенные в классах, называются методами.

В Python функции - это объекты (значения), которые обрабатываются наравне с другими объектами. Таким образом, одну функцию можно передавать в качестве аргумента другой функции во время ее вызова. Точно так же функция может вернуть другую функцию в качестве результате вызова. Функция, как и любой другой объект, может быть связана с переменной, может содержаться в контейнере в качестве элемента и может служить атрибутом объекта. Кроме того, функции могут использоваться в качестве ключей в словарях.

Чаще всего функции определяются с помощью инструкции `def`.

Параметры - это необязательный список идентификаторов, которые связываются со значениями, предоставляемыми при вызове функции; эти значения, чтобы отличить их от параметров, называют аргументами.

Параметры, указанные с помощью простых идентификаторов, - это так называемые позиционные параметры (другое название - обязательные параметры). Соответствующие значения (аргументы) для каждого обязательного (позиционного) параметра должны предоставляться при каждом вызове функции.

````python
def function(x, y):
    pass
````

Вслед за позиционными параметрами (если они предусмотрены для данной функции) можно указать нуль или несколько именованных параметров (другое название - необязательные параметры).

````python
def function(x, y, z=0, w=1):
    pass

y = 1
function(1, y, w=4, z=5)
function(1, 2, 3, 4)
function(x=1, y=2, z=3, w=4)
````

В конце списка параметров можно использовать любую из двух специальных форм: `*args` и `**kwargs`. В этих именах нет ничего особенного - вместо них можно использовать любой идентификатор. Они применяются для передачи в функцию переменного числа аргументов, когда их точное количество заранее неизвестно.

`*args` собирает произвольное количество позиционных аргументов в кортеж.

`**kwargs` собирает произвольное количество именованных аргументов (передаваемых по ключу) в словарь.

````python
def function(*args, **kwargs):
    print(args)
    print(kwargs)

x = 5
function(1, 2, 3, x=x, y="b", z=0)
````

## Классы

Класс - это определяемый пользователем тип, который инстанциализируется для создания экземпляров, т.е. объектов данного типа.

Объекты класса создаются с помощью составной инструкции `class`.

````python
class Class:
    "An empty class."
````

Атрибуты класса - это переменные, которые принадлежат самому классу, а не какому-то конкретному его объекту. Они объявляются прямо внутри тела класса, вне любых методов.

````python
class Class:
    x = 1
    y = "egg"
````

### Экземпляры

Чтобы создать экземпляр класса, необходимо вызвать имя класса, как если бы оно было функцией.

````python
class Class:
    pass

c = Class()  # экземпляр
````

Атрибуты экземпляра (объекта): Личные данные каждого отдельного объекта. Они создаются внутри методов (обычно в `__init__`) с помощью ключевого слова `self`.

Если класс определяет или наследует метод `__init__`, то при вызове объекта класса этот метод автоматически вызывается для нового экземпляра, выполняя специфическую для него инициализацию.

Основным назначением метода `__init__` является создание атрибутов нового экземпляра класса.

````python
class Class:
    def __init__(self, y):
        self.y = y
c1 = Class(1)
c2 = Class(2)
print(c1.y, c2.y)
````

Атрибуты объекта можно просто объявлять, но это не рекомендуется.

````python
class Class():
    pass

c = Class()
c.x = 10    # Проверка типов подсказывает, что атрибут не существует
print(c.x)
````

Атрибуты объекта могут перекрывать атрибуты класса (shadowing):

````python
class Class():
    x = 1

c1 = Class()
c2 = Class()
c2.x = 2     # Shadowing
print(c1.x, c2.x)
````
