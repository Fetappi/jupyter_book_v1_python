```python
class Student:
    """Сущность студента. Отвечает только за свои данные и расчет личного среднего балла."""

    def __init__(self, name, group, grades):
        self.name = name
        self.group = group
        self.grades: list[int] = grades if grades is not None else []

    def add_grade(self, grade):
        """Добавляет оценку или список оценок."""
        if isinstance(grade, int):
            self.grades.append(grade)
        elif isinstance(grade, list):
            self.grades.extend(grade)

    @property
    def average_grade(self):
        """Вычисляет средний балл студента."""
        if not self.grades:
            return 0.0
        return round(sum(self.grades) / len(self.grades), 2)
```

```python

class StudentManager:
    """Менеджер учебной группы. Управляет коллекцией студентов и логикой."""

    def __init__(self):
        self._students = {}

    def add_student(
        self, name, group, grades=None
    ):
        """Добавляет нового студента."""
        self._students[name] = Student(name, group, grades)

    def remove_student(self, name):
        """Удаляет студента по имени."""
        if name in self._students:
            del self._students[name]

    def add_grade(self, name, grade):
        """Добавляет оценку студенту."""
        if name in self._students:
            self._students[name].add_grade(grade)

    def get_average_grade_by_group(self, group):
        """Считает средний балл по заданной группе."""
        group_students = [s for s in self._students.values() if s.group == group]
        if not group_students:
            return 0.0

        total_avg = sum(s.average_grade for s in group_students)
        return round(total_avg / len(group_students), 2)

    def print_all(self):
        """Выводит информацию обо всех студентах."""
        for name, student in self._students.items():
            print(
                f"Студент: {name} | Группа: {student.group} | "
                f"Оценки: {student.grades} | Средний балл: {student.average_grade}"
            )
```

```python
# Инициализация менеджера с поддержкой JSON-хранилища
msu_physics = StudentManager()

# Добавление данных (аналог ДЗ 6)
msu_physics.add_student("Анна", "ФИЗ-101", [4, 5, 4])
msu_physics.add_student("Артем", "ФИЗ-114")
msu_physics.add_student("Павел", "ФИЗ-114", [4, 4, 4])

# Добавление оценок
msu_physics.add_grade("Артем", 5)
msu_physics.add_grade("Артем", [5, 4, 5])

# Расчет среднего балла по группе
avg_group = msu_physics.get_average_grade_by_group("ФИЗ-114")
print(f"Средний балл группы ФИЗ-114: {avg_group}\n")

# Вывод всех студентов
msu_physics.print_all()
```
